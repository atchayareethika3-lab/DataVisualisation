```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Thiranex_Internship/supermarket.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8,5))
df['Sales'].hist(bins=30)
plt.title('Distribution of Sales ')
plt.ylabel('Frequency')
plt.xlabel('Sales')
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales
```

```python
category_sales.plot(kind='bar')

plt.title("Sales by Category")

plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

region_sales.plot(kind='bar')

plt.title("Sales by Region")

plt.show()
```

```python
state_sales = df.groupby('State')['Sales'].sum()

top10 = state_sales.sort_values(ascending=False).head(10)

top10.plot(kind='bar')

plt.title("Top 10 States by Sales")

plt.show()
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

segment_sales.plot(kind='pie', autopct='%1.1f%%')

plt.ylabel('')

plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.show()
```

```python
sns.boxplot(x=df['Sales'])

plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
```

```python
df['Year'] = df['Order Date'].dt.year
```

```python
year_sales = df.groupby('Year')['Sales'].sum()

year_sales.plot(kind='line', marker='o')

plt.title("Sales Trend Over Years")

plt.show()
```
